## 4) Person Detection on an Image
อัปโหลดภาพของคุณเอง (แนะนำ: ไม่ระบุตัวตน / เบลอหน้า)

```python
from google.colab import files
uploaded = files.upload()
img_path = next(iter(uploaded.keys()))
img_path
```

```python
# Detect only 'person' (COCO class 0)
results = model.predict(source=img_path, classes=[0], conf=0.25, device=0, save=True)

import matplotlib.pyplot as plt
img = results[0].plot()
plt.figure(figsize=(10,6))
plt.imshow(img)
plt.axis('off')
plt.show()
```

## 6) Fine-tune YOLOv8 with Public Dataset (COCO8)
ใช้ COCO8 ซึ่งเป็น public dataset ขนาดเล็ก เหมาะกับการสาธิต

```python
from ultralytics.utils.downloads import download
download('https://github.com/ultralytics/assets/releases/download/v0.0.0/coco8.zip')
!unzip -q coco8.zip
!ls
```

```python
# Train (1–3 epochs for demo)
from ultralytics import YOLO

ft_model = YOLO('yolov8n.pt')
ft_model.train(
    data='coco8.yaml',
    epochs=3,
    imgsz=640,
    batch=16,
    device=0
)
```

## 7) Compare Before vs After Training
ใช้ภาพเดียวกันทดสอบโมเดลก่อนและหลัง Fine-tune

```python
# Before
base = YOLO('yolov8n.pt')
base.predict(source=img_path, classes=[0], conf=0.25, device=0, save=True)

# After (adjust path if needed)
tuned = YOLO('runs/detect/train/weights/best.pt')
tuned.predict(source=img_path, classes=[0], conf=0.25, device=0, save=True)
```
